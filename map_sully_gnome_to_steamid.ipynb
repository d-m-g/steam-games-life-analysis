```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import re
import glob 
```

```python
steam = pd.read_csv("data/steam-games-dataset/applicationInformation.csv", encoding="cp1252")
sully = pd.concat([pd.read_csv(f) for f in glob.glob("data/sullygnome/monthly/*.csv")])
games = sully.groupby(["id", "name", "url"], as_index=False)["viewminutes"].sum()
```

```python
mapping = pd.read_csv("data/sullygnome/appid_map.csv")

monthly = pd.concat([
    pd.read_csv(f).assign(month=re.search(r"\d{4}-\d{2}", f).group(0))
    for f in sorted(glob.glob("data/sullygnome/monthly/*.csv"))
])

tw = (monthly.merge(mapping[["appid", "sully_id", "name_steam"]],
                    left_on="id", right_on="sully_id")
             .loc[:, ["appid", "name_steam", "month", "avgviewers", "maxviewers",
                      "viewminutes", "avgchannels", "maxchannels", "uniquechannels"]]
             .sort_values(["appid", "month"])
             .reset_index(drop=True))

tw.to_csv("data/sullygnome/twitch_monthly_by_appid.csv", index=False)
```
